**iii) Define these rules in a separate text file and read them. (Use editfile function in R (package editrules). Use similar function in Python).**

```python
df
```

```python
#first replace target class with text
df['target'] = df['target'].replace({0: 'Iris-setosa', 1: 'Iris-versicolor', 2: 'Iris-virginica'})
```

```python
display (df)
```

```python
print(df.to_string())
```

```python
rule1 = df['target'].apply(lambda x: True if x== 'Iris-setosa' or x== 'Iris-versicolor' or x=='Iris-virginica' else False)
rule2 = df.iloc[:,:4].apply(lambda x: True if all(y>0 for y in x) else False, axis=1)
rule3 = df.apply(lambda x:True if x['petal length (cm)'] >=2*x['petal width (cm)'] else False, axis=1)
rule4 = df['sepal length (cm)'].apply(lambda x: True if x<=30 else False)
rule5 = df.apply(lambda x:True if x['sepal length (cm)'] > x['petal length (cm)'] else False, axis=1)
```

```python
rule1
```

```python
rule3
```

```python
df['petal width (cm)'] = df['petal width (cm)'].replace(0.2, 1.2)
```

```python
display (df)
```

```python
rule4
```

**iv) Determine how often each rule is broken (violatedEdits). Also summarize and plot the result.**

```python
df_rules=pd.DataFrame({"Rule1":rule1, "Rule2":rule2, "Rule3":rule3, "Rule4":rule4, "Rule5":rule5})
df_rules= df_rules.astype(int)
df_rules
```

```python
df_rules["Rule1"].value_counts()
```

```python
df_rules["Rule2"].value_counts()
```

```python
df_rules["Rule3"].value_counts()
```

```python
df_rules["Rule4"].value_counts()
```

```python
df_rules["Rule5"].value_counts()
```

```python
plt.figure()
df_rules.apply(lambda x:len(x) -x.sum()).plot(kind='bar')
plt.xlabel("Rules")
plt.ylabel("Number of records that violate the Rules")
```
