**Q3. Load the data from wine dataset. Check whether all attributes are standardized or not (mean is 0 and standard deviation is 1). If not, standardize the attributes. Do the same with Iris dataset.**

```python
# import the necessary libraries:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import HTML, display
from sklearn.datasets import load_iris
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = StandardScaler() 
scaler1 = MinMaxScaler()
```

```python
iris=load_iris()
df=pd.DataFrame(iris.data, columns=iris.feature_names)
df["Class"] = iris.target
```

```python
df.head()
```

```python
len(df.columns)
```

```python
X=df.iloc[:, :4]
```

```python
X
```

```python
print("Mean:\n",X.mean())
```

```python
print("Standard Deviation:\n",X.std())
```

```python
scaled_iris_data = scaler.fit_transform(X)
```

```python
scaled_iris_data = pd.DataFrame(scaled_iris_data, columns=X.columns)
```

```python
scaled_iris_data.head

#StandardScaler standardizes features by removing the mean and scaling to unit variance.
#It's useful when your features have different units or when you want to make features comparable.
```

```python
scaled_iris_data1 = scaler1.fit_transform(X)
scaled_iris_data1 = pd.DataFrame(scaled_iris_data1, columns=X.columns)
scaled_iris_data1.head

#MinMaxScaler scales features to a given range (default is [0, 1]).
#It's useful when you need values in a specific range, like for neural networks with sigmoid activation functions.
```
